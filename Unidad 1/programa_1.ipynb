```python
tabla = int (input("¿Qué tabla quieres? "))
lim1 = int (input("¿Cuál es el limite inicial? "))
lim2 = int (input("¿Cuál es el limite final? "))
inte = int (input("¿Cuál es el intervalo? "))

# agregamos la instruccion while para que valide que el valor iniciar sea mayor que el final, de lo contrario me pedirá indefinidamente el valor final hasta que sea mayor
while lim1 >= lim2:
  print("El limite final debe ser mayor que el inicio")
  lim2 = int(input ("¿Cuál es el limite final? "))

for i in range(lim1, lim2, inte):
  print(tabla, "*", i, "=", tabla*i)
```

```python
# Podemos acceder a cada elemento o item de la lista a taves del indice o su posicion, las listas siempre empiezan en la posicion 0
alumnos = ["Ana", "Luis", "Carlos"]
print(alumnos)
print(alumnos[0])
print(alumnos[2])
```

```python
#podemos modificar el valor de un elemento de la lista usando el nombre de la lista y el indice y le asignamos el nuevo valor
print(alumnos)
alumnos[1]="Juan"
print(alumnos)
```

```python
# podemos especificar indices negativos y cuandi se ponen valores negativos se lee o se accede a la lista desde el final empezando con -1
print(alumnos)
print(alumnos[-1])
print(alumnos[-2])
```

```python

#El metodo append agrega un elemento al final de la lista
print(alumnos)
alumnos.append("Sandra")
print(alumnos)
```

```python
print(alumnos)
alumnos.insert(2, "Sofía")
print(alumnos)
```

```python
#remove borra un elemento de la lista especificando el valor, si existen valores repetidos, borra el primer valor que encuentra
print(alumnos)
alumnos.remove("Sofía")
print(alumnos)
```

```python
print(alumnos[2])
#la siguiente linea marca error porque queremos acceder a un indice que no existe
#la lista tiene unicamente 4 elementos y queremos acceder al 7 que no existe
print(alumnos[7])
```

```python
#pop me permite borrar un elemento de la lista especificando su posicion
print(alumnos)
alumnos.pop(1)
print(alumnos)
```

```python
#del me permite borrar un elemento de la lista especificando su posicion
print(alumnos)
del alumnos[1]
print(alumnos)
```

```python
# la funcion len me permite conocer el numero de elementos de la lista
print("Numero de elementos o items en la lista", len(alumnos))
```
